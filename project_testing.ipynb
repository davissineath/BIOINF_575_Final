```python
import csv
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
def read_chip(path): 
    df = pd.read_csv(
            path, 
            sep='\t', 
            names = ['CHR', 'POS', 'dbSNP_ID', 'ALLELE'],
            header=None
            )
    df = df.rename(columns={'ALLELE': 'ALLELE_23andme'})
    return df

def read_pharmgkb(path): 
    df = pd.read_csv(
            path, 
            sep ='\t',
            quoting=csv.QUOTE_NONE
            )
    df = df.rename(columns={'Alleles': 'ALLELE_PharmGKB'})
    return df

def split_column(df, col):
    df[col] = df[col].str.split(',')
    df = df.explode(col)
    return df

def merge_chip_pharmgkb(chip, pgx): 
    merged_df = pd.merge(
                chip, 
                pgx, 
                left_on="dbSNP_ID", 
                right_on="Variant/Haplotypes", 
                how="inner"
                )

    # Cols to keep: 
    # dbSNP_ID, GENE_SYMBOL, DRUG_NAME, PMID, PHENOTYPE_CATEGORY, SIGNIFICANCE, 
    # ASSOCIATED, DIRECTION, NOTES, SENTENCE, ALLELE_PharmGKB, ALLELE_23andme

    cols_to_keep = [
        'dbSNP_ID', 
        'Gene',
        'Drug(s)', 
        'PMID', 
        'Phenotype Category', 
        'Significance', 
        'Is/Is Not associated', 
        'Direction of effect', 
        'Notes', 
        'Sentence', 
        'ALLELE_PharmGKB', 
        'ALLELE_23andme'
    ]
    
    return merged_df.loc[:, cols_to_keep]
```

```python
# Chip Table
chip = read_chip('data/23andme_v5_hg19_ref.txt.gz')
display(chip.head())

# Annotation Table
pharmgkb = read_pharmgkb('data/var_drug_ann.tsv')
display(pharmgkb.head())

# One row per variant
vars_split_pharmgkb = split_column(pharmgkb, 'Variant/Haplotypes')
display(vars_split_pharmgkb.head())

# merged table
merged = merge_chip_pharmgkb(chip, vars_split_pharmgkb)
display(merged.head())

# merged table one row per drug
# **uncomment below if you need one row per drug** #
# merged = split_column(merged, 'Drug(s)') 
# display(merged.head())
```
